```python
import pandas as pd

# Load the data
df = pd.read_csv('../data/bank_reviews.csv')
```

```python
df. describe()
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df['date'] = pd.to_datetime(df['date']).dt.strftime('%Y-%m-%d')
```

```python
df = df.dropna()
```

```python
df.to_csv('../data/processed_bank_reviews.csv', index=False)
print("Processed data saved to data/processed_bank_reviews.csv")
```
